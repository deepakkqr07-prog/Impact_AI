## Reading the Files

```python
def read_text(path):
    
    with open(path, 'r') as file:
        
        lines = file.readlines()
    
    # Now lines is a list where each element is a line from the file
    text = []

    for line in lines:
        text.append(line)

    df = pd.DataFrame()
    
    if path.find('emotion') != -1:
        
        df['emotion'] = text
        
    elif path.find('act') != -1:
        
        df['act'] = text
        
    else:
        
        df['text'] = text
        
        
    return df
```

```python
train_text = read_text('/kaggle/input/contextual-1/dialogues_train.txt')
valid_text = read_text('/kaggle/input/contextual-1/dialogues_validation.txt')
test_text = read_text('/kaggle/input/contextual-1/dialogues_test.txt')

train_emotion = read_text('/kaggle/input/contextual-1/dialogues_emotion_train.txt')
valid_emotion = read_text('/kaggle/input/contextual-1/dialogues_emotion_validation.txt')
test_emotion = read_text('/kaggle/input/contextual-1/dialogues_emotion_test.txt')

train_act = read_text('/kaggle/input/contextual-1/dialogues_act_train.txt')
valid_act = read_text('/kaggle/input/contextual-1/dialogues_act_validation.txt')
test_act = read_text('/kaggle/input/contextual-1/dialogues_act_test.txt')
```

```python
train = pd.concat([train_text, train_act, train_emotion], axis = 1)
train
```

```python
valid = pd.concat([valid_text, valid_act, valid_emotion], axis = 1)
valid
```

```python
test = pd.concat([test_text, test_act, test_emotion], axis = 1)
test
```

## Cleaning the files

```python
def clean_dataframe(df):
    """
    This function takes a pandas DataFrame with 'text', 'act', and 'emotion' columns and:
    - Replaces '__eou__' with an empty string in the 'text' column
    - Replaces '\n' with an empty string in 'act' and 'emotion' columns

    Parameters:
    - df (pd.DataFrame): The input DataFrame to clean.

    Returns:
    - pd.DataFrame: The cleaned DataFrame.
    """
    df['text'] = df['text'].str.replace('__eou__', '', regex=False)
    df['act'] = df['act'].str.replace('\n', '', regex=False)
    df['emotion'] = df['emotion'].str.replace('\n', '', regex=False)

    
clean_dataframe(train)
clean_dataframe(valid)
clean_dataframe(test)
```

# Training

```python
import pandas as pd
import torch
from torch.utils.data import DataLoader, TensorDataset
from transformers import BertTokenizer, BertForSequenceClassification, AdamW
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
```

```python
model = BertForSequenceClassification.from_pretrained('bert-base-uncased')
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# Function to encode the dataset
def encode_dataset(df):
    input_ids = []
    attention_masks = []
    labels = []

    for _, row in df.iterrows():
        encoded_dict = tokenizer.encode_plus(
                            row['text'],                      
                            add_special_tokens = True, 
                            max_length = 64,           
                            pad_to_max_length = True,
                            return_attention_mask = True,   
                            return_tensors = 'pt',     
                       )
        
        input_ids.append(encoded_dict['input_ids'])
        attention_masks.append(encoded_dict['attention_mask'])
        labels.append(row['emotion']) # Assuming 'emotion' is a column in your dataframe

    input_ids = torch.cat(input_ids, dim=0)
    attention_masks = torch.cat(attention_masks, dim=0)
    labels = torch.tensor(labels)

    return TensorDataset(input_ids, attention_masks, labels)

# Encode the datasets
train_dataset = encode_dataset(train)
valid_dataset = encode_dataset(valid)
test_dataset = encode_dataset(test)

# Create DataLoaders
batch_size = 32

train_dataloader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
validation_dataloader = DataLoader(valid_dataset, batch_size=batch_size)
test_dataloader = DataLoader(test_dataset, batch_size=batch_size)
```

```python
# Initialize the model
model = BertForSequenceClassification.from_pretrained(
    'bert-base-uncased',
    num_labels = 7,  # Assuming 7 emotions including 'no emotion'
    output_attentions = False,
    output_hidden_states = False,
)

# Move the model to GPU if available
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)

# Optimizer and learning rate scheduler
optimizer = AdamW(model.parameters(), lr=2e-5)
```

```python
# Training loop
for epoch in range(1):  # Number of training epochs
    model.train()
    total_loss = 0

    for step, batch in enumerate(train_dataloader):
        b_input_ids = batch[0].to(device)
        b_input_mask = batch[1].to(device)
        b_labels = batch[2].to(device)

        model.zero_grad()        

        outputs = model(b_input_ids, 
                        token_type_ids=None, 
                        attention_mask=b_input_mask, 
                        labels=b_labels)
        
        loss = outputs.loss
        total_loss += loss.item()

        loss.backward()
        optimizer.step()

    avg_train_loss = total_loss / len(train_dataloader)            

    print(f'Average training loss: {avg_train_loss}')

# Evaluation function
def evaluate(dataloader):
    model.eval()
    predictions, true_labels = [], []

    for batch in dataloader:
        batch = tuple(t.to(device) for t in batch)
        b_input_ids, b_input_mask, b_labels = batch

        with torch.no_grad():
            outputs = model(b_input_ids, token_type_ids=None, 
                            attention_mask=b_input_mask)

        logits = outputs.logits
        logits = logits.detach().cpu().numpy()
        label_ids = b_labels.to('cpu').numpy()
        
        predictions.extend(logits)
        true_labels.extend(label_ids)

    return predictions, true_labels

# Evaluate on the validation set
predictions, true_labels = evaluate(validation_dataloader)
```
